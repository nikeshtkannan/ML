```python
# Experiment 8: K-Nearest Neighbours using Fashion MNIST

import pandas as pd
import time
import matplotlib.pyplot as plt
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score

# 1. Load Fashion MNIST dataset
train_data = pd.read_csv("fashion-mnist_train.csv")
test_data = pd.read_csv("fashion-mnist_test.csv")

# 2. Separate features and labels
X_train = train_data.drop("label", axis=1)
y_train = train_data["label"]

X_test = test_data.drop("label", axis=1)
y_test = test_data["label"]

# 3. Use a smaller portion to reduce computation time
X_train = X_train.iloc[:10000]
y_train = y_train.iloc[:10000]

X_test = X_test.iloc[:2000]
y_test = y_test.iloc[:2000]

# 4. Normalize pixel values
X_train = X_train / 255.0
X_test = X_test / 255.0

# 5. Different values of K
k_values = [1, 3, 5, 7, 9, 11]

results = []

# 6. Train and test KNN for each K
for k in k_values:

    print("\nK =", k)

    start_time = time.time()

    # Create KNN classifier
    knn = KNeighborsClassifier(n_neighbors=k)

    # Train ty_train = train_data["label"]
he model
    knn.fit(X_train, y_train)

    # Predict test data
    y_pred = knn.predict(X_test)

    end_time = time.time()

    # Calculate accuracy
    accuracy = accuracy_score(y_test, y_pred)

    # Calculate computation time
    computation_time = end_time - start_time

    results.append((k, accuracy, computation_time))

    print("Accuracy:", accuracy)
    print("Computation Time:", computation_time, "seconds")


# 7. Display comparison
print("\n----- KNN Performance Comparison -----")
print("K\tAccuracy\tComputation Time")

for k, accuracy, computation_time in results:
    print(k, "\t", round(accuracy, 4),
          "\t\t", round(computation_time, 2), "seconds")


# 8. Graph
k_values = [result[0] for result in results]
accuracies = [result[1] for result in results]

plt.plot(k_values, accuracies, marker='o')
plt.xlabel("K Value")
plt.ylabel("Accuracy")
plt.title("KNN Accuracy vs K Value")
plt.xticks(k_values)
plt.grid(True)
plt.show()
```

```python
# 9. Graph: K Value vs Computation Time
computation_times = [result[2] for result in results]

plt.plot(k_values, computation_times, marker='o')
plt.xlabel("K Value")
plt.ylabel("Computation Time (seconds)")
plt.title("KNN Computation Time vs K Value")
plt.xticks(k_values)
plt.grid(True)
plt.show()
```
